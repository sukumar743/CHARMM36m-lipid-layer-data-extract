```
import pandas as pd
import os

train_file_path = 'train-data.tsv'

# Check if the file exists, if not, download it using wget
if not os.path.exists(train_file_path):
    print("Downloading training data...")
    os.system("wget -q https://cdn.freecodecamp.org/project-data/sms/train-data.tsv")

# Load the dataset using tsv delimiter, adding column names 'label' and 'message'
df_train = pd.read_csv(train_file_path, sep='\t', names=['label', 'message'])
display(df_train.head())
```

```
import tensorflow as tf
import numpy as np
from tensorflow.keras.layers import TextVectorization, Embedding, Bidirectional, LSTM, Dense, Dropout

# Load validation dataset
valid_file_path = 'valid-data.tsv'
import os
if not os.path.exists(valid_file_path):
    os.system("wget -q https://cdn.freecodecamp.org/project-data/sms/valid-data.tsv")

df_valid = pd.read_csv(valid_file_path, sep='\t', names=['label', 'message'])

# Map labels to binary values (0 for ham, 1 for spam)
df_train['label_num'] = df_train['label'].map({'ham': 0, 'spam': 1})
df_valid['label_num'] = df_valid['label'].map({'ham': 0, 'spam': 1})

# Parameters
VOCAB_SIZE = 10000
MAX_LEN = 100

# Vectorize the text data
vectorizer = TextVectorization(max_tokens=VOCAB_SIZE, output_sequence_length=MAX_LEN)
vectorizer.adapt(tf.constant(df_train['message'].values.tolist(), dtype=tf.string))

# Build a more robust model using a Bidirectional LSTM
model = tf.keras.Sequential([
    vectorizer,
    Embedding(VOCAB_SIZE, 64, mask_zero=True),
    Bidirectional(LSTM(32)),
    Dense(32, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model
model.fit(
    tf.constant(df_train['message'].values.tolist(), dtype=tf.string),
    np.array(df_train['label_num'].values, dtype=np.float32),
    epochs=10,
    validation_data=(
        tf.constant(df_valid['message'].values.tolist(), dtype=tf.string),
        np.array(df_valid['label_num'].values, dtype=np.float32)
    )
)
```
